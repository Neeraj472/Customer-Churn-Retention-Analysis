```python
df = df.apply(
    lambda x: x.str.strip() if x.dtype == "object" else x
)
```

```python
df.columns
```

```python
cols=["Customer_Name","Gender","State","City","Subscription_Type","Payment_Method","Contract_Type","Internet_Service","Churn","Dependents","Senior_Citizen","Tech_Support"]
for i in cols:
    df[i]=df[i].str.title()
```

```python
df.head(5)
```

```python
df.replace(["NaN","N/A",""," "],np.nan,inplace=True)
```

```python
df["Churn"].value_counts()
```

```python
df.info()
```

```python
cols=["Age","Tenure_Months","Monthly_Charges","Total_Charges"]
for i in cols:
    df[i]=pd.to_numeric(df[i],errors="coerce")
```

```python
df.describe()
```

```python

df = df[(df["Age"]>=18) &(df["Age"]<=100)]
# remove negative values
df = df[df["Monthly_Charges"]>=0]
df = df[df["Total_Charges"]>=0]
```

```python
df.isnull().sum()
```

```python
df["Customer_value"]=df["Monthly_Charges"]*df["Tenure_Months"]
```

```python
df.head(2)
```

```python
df["Monthly_Revenue"]=df["Monthly_Charges"]
```

```python
#Tenure grouping "0-12" "13-24" "25-48" "49-72"
bins =[0,12,24,48,72]
labels =["0-12 month","13-24 month","25-48 month","49-72 month"]
df["Tenure_group"]=pd.cut(df["Tenure_Months"],bins=bins,labels=labels)
```

```python
df["Senior_Flag"]=np.where(df["Age"]>=60,"Senior","Adult")
```

```python
df["Churn_flag"]=df["Churn"].map({"Yes":1,"No":0})
```
